### Conclusions and findings / Conclusiones y hallazgos

**[EN]**

The analysis of the annual crime datasets revealed several characteristics that will guide the subsequent data preparation and modeling stages. Since each record represents an investigation case file and its associated crime, only the variables directly related to the crime event—namely the occurrence date, occurrence time, and spatial coordinates (X and Y)—will be retained for training the STKDE model.

Based on this finding, two final datasets will be created. The first dataset will contain only the spatio-temporal variables required by the STKDE algorithm to estimate crime risk and compute spatial and temporal distances. The second dataset will preserve additional attributes, such as crime type, investigation start date, and other administrative information, to support descriptive statistics, exploratory data analysis (EDA), and crime characterization.

Although each annual CSV file is organized according to the investigation start date, the crime occurrence date does not always fall within the corresponding calendar year. Therefore, only records whose crime occurrence date lies within the study period (2020–2025) will be retained to ensure temporal consistency throughout the analysis.

Finally, the datasets include a wide variety of gender-related crimes. However, the final training dataset will include only those offenses that can reasonably occur in public spaces or during everyday mobility. Crimes such as domestic violence may involve contextual and interpersonal factors that differ substantially from crimes committed in public environments. Consequently, each crime category will be evaluated according to its relationship with the surrounding environment before determining whether it should be included in the final dataset.

**[ES]**

El análisis de los datasets anuales permitió identificar diversas características que servirán como base para las siguientes etapas de preparación de datos y construcción del modelo. Debido a que cada registro representa una carpeta de investigación y el delito asociado a ella, para el cálculo de las distancias dentro del modelo STKDE, únicamente se conservarán las variables directamente relacionadas con el evento delictivo, principalmente la fecha de los hechos, la hora de los hechos y las coordenadas espaciales (X y Y).

Derivado de este hallazgo, se construirán dos conjuntos de datos finales. El primero contendrá exclusivamente las variables espacio-temporales necesarias para alimentar el modelo STKDE y realizar el cálculo de distancias espaciales y temporales. El segundo conservará atributos adicionales, como el tipo de delito, la fecha de inicio de la carpeta de investigación y otras variables, con el propósito de realizar el análisis exploratorio de datos (EDA), generar estadísticas descriptivas y caracterizar la incidencia delictiva.

Aunque cada archivo anual se encuentra organizado de acuerdo con la fecha de inicio de la carpeta de investigación, se identificó que la fecha de ocurrencia de los hechos no siempre pertenece al mismo periodo. Por ello, con el fin de mantener la consistencia temporal del estudio, únicamente se conservarán aquellos registros cuya fecha del delito se encuentre dentro del periodo de análisis comprendido entre 2020 y 2025.

Finalmente, los conjuntos de datos incluyen una amplia variedad de delitos con perspectiva de género. Sin embargo, para la construcción del conjunto de datos definitivo se priorizarán aquellos delitos que puedan experimentarse en espacios públicos o durante los desplazamientos cotidianos de las usuarias. Delitos como la violencia familiar pueden estar influenciados por factores propios del entorno familiar y no necesariamente por las condiciones del espacio público. Por esta razón, cada categoría delictiva será analizada individualmente para determinar su pertinencia dentro del modelo de estimación de riesgo espacio-temporal.

**[EN]**

However, since the STKDE model estimates the spatio-temporal density of crime events rather than investigation case files, it is necessary to determine whether multiple investigation records may refer to the same criminal event.

This verification is required because, if the same crime event is represented multiple times in the training dataset, the STKDE algorithm would interpret them as independent occurrences, artificially increasing the estimated density at that location and time, which could introduce bias into the resulting risk surface. Nevertheless, it is also acknowledged that situations may exist where different investigation case files legitimately share the same crime characteristics, such as multiple victims involved in the same incident, independent reports of an event, or multiple offenses derived from a single criminal act.

To identify these potential duplicate events, the analysis considered only the attributes describing the criminal incident itself, excluding administrative information associated with the investigation case. The following variables were therefore selected:

- 'TIPO IMPACTO'
- 'DELITO'
- 'MODALIDAD - DELITO'
- 'FECHA DE LOS HECHOS'
- 'HORA DE LOS HECHOS'
- 'COLONIA HECHOS'
- 'ALCALDÍA HECHOS'
- 'COORD. X'
- 'COORD. Y'

**[ES]**

Sin embargo, dado que el modelo Spatio-Temporal Kernel Density Estimation (STKDE) estima la densidad espacio-temporal de eventos delictivos y no de carpetas de investigación, resulta necesario verificar si existen diferentes carpetas de investigación que potencialmente hagan referencia a un mismo hecho delictivo.

Esta verificación es necesaria porque, si un mismo evento delictivo se encuentra representado más de una vez dentro del conjunto de datos de entrenamiento, el algoritmo STKDE lo interpretará como múltiples ocurrencias independientes en una misma ubicación y momento, incrementando artificialmente la densidad estimada en esa región e introduciendo un posible sesgo en la superficie de riesgo obtenida. No obstante, también es importante reconocer que pueden existir situaciones legítimas en las que diferentes carpetas de investigación compartan las mismas características del hecho delictivo, por ejemplo, cuando un mismo incidente involucra a múltiples víctimas, cuando existen denuncias independientes de un mismo evento o cuando una misma situación da origen a varios delitos relacionados.

Con el objetivo de identificar estos posibles eventos duplicados, el análisis se realizó considerando únicamente las variables que describen el hecho delictivo, excluyendo la información administrativa propia de cada carpeta de investigación. Para ello, se seleccionaron los siguientes atributos:

- 'TIPO IMPACTO'
- 'DELITO'
- 'MODALIDAD - DELITO'
- 'FECHA DE LOS HECHOS'
- 'HORA DE LOS HECHOS'
- 'COLONIA HECHOS'
- 'ALCALDÍA HECHOS'
- 'COORD. X'
- 'COORD. Y'

**[EN]**

The results indicate that the proportion of potential duplicate events ranges between 0.04% and 0.18% of the total records, suggesting that this situation is relatively uncommon throughout the available historical data.

Considering both the characteristics of the STKDE algorithm and the low proportion of affected records, these entries will be treated as duplicate representations of the same event for modeling purposes. Consequently, only the investigation case file with the most recent investigation start date will be retained. This criterion assumes that the latest record corresponds to the most up-to-date administrative representation of the event while preventing multiple representations of the same crime from artificially increasing the estimated spatio-temporal density.

It is important to note that this deduplication procedure will be applied exclusively to the dataset used for training the STKDE model. The complete administrative dataset, including all investigation case files, will be preserved for descriptive analyses and statistical reporting, ensuring that no institutional information is lost during the data preparation process.

**[ES]**

Los resultados muestran que la proporción de posibles eventos delictivos duplicados varía entre 0.04 % y 0.18 % del total de registros, lo que indica que este fenómeno representa una fracción muy reducida de la información disponible.

Considerando la naturaleza del algoritmo STKDE y el reducido porcentaje de registros involucrados, como decisión metodológica dentro del proceso de preparación de datos para la estimación de densidad, estos registros serán tratados como representaciones duplicadas de un mismo evento delictivo. En consecuencia, únicamente se conservará la carpeta de investigación cuya fecha de inicio sea la más reciente, bajo el supuesto de que corresponde a la representación administrativa más actualizada del hecho, evitando así que un mismo evento contribuya múltiples veces al cálculo de la densidad espacio-temporal.

Es importante señalar que este procedimiento de depuración se aplicará exclusivamente al conjunto de datos destinado al entrenamiento del modelo STKDE. El conjunto de datos administrativo original, que conserva la totalidad de las carpetas de investigación, permanecerá sin modificaciones y será utilizado para el análisis descriptivo, la caracterización de la incidencia delictiva y la generación de estadísticas, garantizando así que no se pierda información institucional relevante.

In the dataset of investigation files with a gender perspective in Mexico City, a proportion of records shows the value "SIN REGISTRO" (NO RECORD) in the coordinate columns (COORD. X and COORD. Y). However, in many of those cases, partial spatial information is available in other variables: street(s), neighborhood (colonia), and borough (alcaldía) of the place of the events.

Given that the analysis requires locating the events in geographical space, it was not feasible to immediately discard all records without coordinates. At the same time, it was also not appropriate to assume that all textual address information is reliable: the data contains values that do not correspond to real locations (for example, "SIN REGISTRO", "DESCONOCIDO" [UNKNOWN], "NO ESPECIFICA LUGAR DE LOS HECHOS" [DOES NOT SPECIFY PLACE OF EVENTS], or narrative texts that describe the situation rather than a physical address).

Therefore, a selective cleaning and geocoding process was defined, with the following criteria:

1. Identify the records with missing coordinates ("SIN REGISTRO").

2. Construct an address by concatenating street, neighborhood, and borough in a format compatible with geocoding services in Mexico.

3. Exclude non-geocodable components before querying any API.

4. Query the Google Maps Geocoding API only when the record met the minimum quality criterion: at least one valid street, neighborhood, and borough.

5. Keep the records that did not meet this criterion unmodified, to avoid introducing inaccurate or artificial coordinates.

This decision responds to a methodological balance: recovering useful spatial information without compromising the quality of the analysis. Automatic geocoding allows leveraging data already captured in the police record, but only when there is a sufficiently structured address to obtain a reasonable location.

The results of the process confirm that this strategy was conservative and appropriate. This indicates that the absence of coordinates in most cases is not solely due to a technical failure, but rather because the registered address information is incomplete or of low quality.

Consequently, the geocoding process is conceived not as a universal solution to impute all missing coordinates, but as a controlled spatial enrichment and cleaning stage, aimed at:

- Maximizing the use of records with complete addresses.

- Avoiding assigning coordinates to records with insufficient information.

- Explicitly documenting which records were geocoded and which remain without coordinates.

This approach strengthens the methodological transparency of the thesis and allows distinguishing, in the subsequent analysis, between original coordinates from the dataset, coordinates obtained through geocoding, and records that remain without a reliable spatial location.

##### <font color='#266D00'>Implementation of the geocoding script / Implementación del script de geocodificación</font>

**[EN]**

To automate the recovery of coordinates in records with the value "SIN REGISTRO" (NO RECORD), a Python script was developed that integrates the pandas (dataset management), googlemaps (querying the Google Maps Geocoding API), python-dotenv (secure loading of the API key), and openpyxl (exporting the final file in Excel) libraries.

The general flow of the script is as follows:

1. **Dataset loading (CSV files per year)** and preparation of the coordinate columns.

2. **Filtering of candidate records:** only rows with "SIN REGISTRO" in COORD. X or COORD. Y that also have a valid street, neighborhood, and borough are selected.

3. **Cleaning of address components:** before building the search string, the script discards values that do not represent real locations (for example, "SIN REGISTRO", "DESCONOCIDO", "NO ESPECIFICA LUGAR DE LOS HECHOS", or narrative texts).

4. **Address construction:** valid components are concatenated into a standard format for Mexico, for example:
{street} and {street 2}, {neighborhood}, {borough}, Mexico City, Mexico.

5. **API Query:** for each eligible record, the address is sent to the Google Maps Geocoding API, with a geographical restriction to Mexico (country=MX, region=mx), and the latitude and longitude in the WGS84 system are obtained.

6. **Dataset update:** the recovered coordinates are assigned to COORD. Y (latitude) and COORD. X (longitude), respecting the convention of the original dataset.

7. **Exporting the result:** The final file is saved in CSV format (.csv). Additionally, two more columns were added to the final dataset (`ENVIADO_API` / `SENT_API` and `estado` / `state`). The first column properly identifies the records that were sent to the geocoding API, while the state column indicates whether the coordinates were verified as correct during the manual review.


**Note: The script used can be found in the src/ folder.**

##### <font color='#266D00'>Manual verification of geographic information / Verificación manual de la información geográfica</font>

**[EN]**

Manual verification of geographic information
Although automated geocoding speeds up the process, it does not guarantee on its own that the returned location faithfully corresponds to the place of the events. The API interprets the address text and returns the result it considers most likely, which can generate:

- Approximate locations (center of a neighborhood or borough).

- Matches with ambiguous names (streets named after states, people, or places).

- Results outside the study area when the text points to another federal entity.

Therefore, and given that the number of geocoded records was small (24 in total between 2020 and 2025), a manual case-by-case verification was performed to validate the consistency between the constructed address and the obtained coordinates.

The verification consisted of:

1. **Reviewing the address assembled by the script** (street, neighborhood, and borough) and contrasting it with the original values of the record.

2. **Checking the coordinates in Google Maps** to confirm that the point falls within Mexico City and in an area compatible with the declared neighborhood and borough.

3. **Evaluating the spatial result:** If the location matches the street and neighborhood, is situated in the correct borough, and shows no obvious displacement to another entity, the value "CORRECTO" was assigned to the estado column.

4. **Identifying atypical or erroneous cases**, such as records whose "streets" are state names and the API returns coordinates outside of CDMX (for example, Guanajuato instead of Gustavo A. Madero). In these cases, the value "INCORRECTO" was assigned to the estado column for different treatment in subsequent cleaning steps.

**NOTE: The resulting dataset of this manual verification is saved in data/geocoding and will be used in the next steps of the cleaning process.**
