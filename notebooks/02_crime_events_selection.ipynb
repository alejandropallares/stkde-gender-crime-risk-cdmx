## <font color='#1083D6'>Definition of Selection Criteria / Definición de los criterios de selección</font>

**[EN]**

The selection of crime events was guided by the objective of this project, which is to estimate the spatio-temporal risk of gender-related crimes that women may experience in a specific place and time within Mexico City. Therefore, only crimes whose occurrence can be reasonably associated with characteristics of the surrounding urban environment and whose spatial location represents the actual place where the event occurred were considered.

### Inclusion criteria
Crime records were included when they met the following conditions:
- The offense represents a form of gender-based violence that may occur in public or semi-public spaces.
- The occurrence of the crime may be influenced by environmental, spatial, or temporal characteristics, making it suitable for hotspot analysis.
- The event can contribute to identifying areas with increased risk for women through historical spatio-temporal patterns.

### Exclusion criteria
Crime records were excluded when one or more of the following conditions were met:
- The crime primarily results from interpersonal or domestic relationships rather than characteristics of the surrounding environment (e.g., domestic violence).
- The offense mainly occurs in digital environments or lacks a meaningful physical location (e.g., crimes against sexual privacy).
- The event cannot be directly associated with environmental factors that influence the occurrence of crime in public spaces.

### Methodological justification
The STKDE method estimates crime risk by identifying spatial and temporal concentrations of historical events. Consequently, including crimes whose occurrence is weakly related to the physical environment could introduce spatial bias and reduce the interpretability of the estimated risk surface. For this reason, only offenses whose occurrence is directly associated with specific geographic locations and whose spatial distribution reflects environmental opportunity patterns were selected for model training.

La selección de los eventos delictivos se realizó con base en el objetivo del presente proyecto, el cual consiste en estimar el riesgo espacio-temporal de delitos de género que una mujer puede experimentar en un lugar y momento específicos dentro de la Ciudad de México. En consecuencia, únicamente se consideraron aquellos delitos cuya ocurrencia puede asociarse razonablemente con las características del entorno urbano y cuya ubicación geográfica representa el sitio donde ocurrieron los hechos. 

### Criterios de inclusión
Se incluyeron aquellos registros que cumplieron con las siguientes condiciones:
- El delito constituye una forma de violencia de género susceptible de ocurrir en espacios públicos o semipúblicos.
- La ocurrencia del delito puede verse influenciada por factores espaciales o temporales del entorno, permitiendo la identificación de patrones de concentración.
- El evento aporta información útil para estimar zonas con mayor riesgo para la población femenina a partir del comportamiento histórico de los delitos.

### Criterios de exclusión
Se excluyeron aquellos registros que cumplieron con alguna de las siguientes características:
- El delito se origina principalmente por relaciones personales, familiares o domésticas y no por condiciones propias del entorno urbano (por ejemplo, violencia familiar).
- El delito ocurre principalmente en medios digitales o carece de una ubicación física claramente asociada al evento (por ejemplo, delitos contra la intimidad sexual).
- La ocurrencia del delito no puede relacionarse directamente con factores ambientales que influyan en la seguridad de los espacios públicos.

### Justificación metodológica
El método STKDE estima el riesgo mediante la concentración espacio-temporal de eventos históricos. Por ello, incorporar delitos cuya ocurrencia no depende del entorno físico podría introducir sesgos espaciales y afectar la interpretación de la superficie de riesgo generada. En consecuencia, el conjunto de entrenamiento se conformó únicamente por delitos cuya distribución espacial refleja patrones asociados al entorno urbano y que permiten modelar de manera consistente el riesgo de ocurrencia en el espacio y el tiempo.
