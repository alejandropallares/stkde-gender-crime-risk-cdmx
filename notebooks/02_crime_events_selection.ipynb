# 02. Crime Events Selection / Selección de eventos delictivos

## Objective / Objetivo

**[EN]**

The purpose of this notebook is to build the dataset that will be used as input for the Spatio-Temporal Kernel Density Estimation (STKDE) model.

Starting from the previously cleaned and consolidated dataset, the methodological criteria defined for this research are applied to select only those crime events whose occurrence can be represented as a geographic point and whose spatial dynamics are suitable for spatio-temporal risk analysis in public spaces.

No additional data cleaning or transformation procedures are performed in this notebook. Instead, the dataset is filtered according to the study criteria. The output of this notebook is a dataset containing only the crime events considered for the modeling stage.

**[ES]**

Este cuaderno tiene como objetivo construir el conjunto de datos que será utilizado como entrada para el modelo de estimación de riesgo espacio-temporal (STKDE).

A partir del conjunto de datos consolidado y previamente depurado, se aplican los criterios metodológicos definidos para seleccionar únicamente aquellos delitos cuya ocurrencia puede representarse mediante un punto geográfico y cuya dinámica espacial resulta adecuada para el análisis de riesgo en espacios públicos.

Durante este proceso no se realizan tareas adicionales de limpieza o transformación de datos; únicamente se filtran los registros de acuerdo con los criterios establecidos para el estudio. El resultado de este cuaderno es un conjunto de datos que contiene exclusivamente los eventos delictivos considerados para la etapa de modelado.
